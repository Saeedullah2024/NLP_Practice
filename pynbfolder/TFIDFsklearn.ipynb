```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
    'I love to play football and basketball.',
    'I enjoy watching movies and reading books.',
    'This is the fourth document.',
    'Imran Khan is the former Prime Minister of Pakistan.',
    'Saeedullah is a software engineer and data scientist.',
    'I love to have tea and coffee in the morning.',
]
```

```python
v = TfidfVectorizer()
```

```python
x = v.fit_transform(corpus)
```

```python
print(v.vocabulary_)
```

```python
all_features = v.get_feature_names_out()
```

```python
all_features #This is object its not a list so we can not use index on it. We can convert it to list and then use index on it.
```

```python
for word in all_features:
    indx = v.vocabulary_.get(word) #IT contain every single word and its index in the vocabulary
    print(f"{word} : {v.idf_[indx]}") #Print the IDF value for each word
```

```python
dataf["Description"] = dataf.iloc[:, 1:].fillna("").astype(str).agg(" ".join, axis=1)
```

```python
dataf = dataf.iloc[:, [0, -1]]
```

```python
print(dataf.shape)
print(dataf.head())
```

```python
dataf.columns = ["Category", "Description"]
```

```python
dataf['Category'].value_counts()
```

```python
valid_categories = [
    "Household",
    "Books",
    "Electronics",
    "Clothing & Accessories"
]

dataf = dataf[dataf["Category"].isin(valid_categories)].copy()
```

```python
dataf['Category'].value_counts()
```

```python
dataf.shape
```

```python
dataf["category_label_encode"] = dataf["Category"].map({
    "Household": 0,
    "Books": 1,
    "Electronics": 2,
    "Clothing & Accessories": 3
})
```

```python
dataf.head(5)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train , y_test = train_test_split(dataf['Description'] , dataf['category_label_encode'], test_size=0.20 , stratify=dataf['category_label_encode'])
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
clf = Pipeline([
    ('TfVectorizer' , TfidfVectorizer()),
    ('KnClass' , KNeighborsClassifier())
])
```

```python
clf.fit(X_train , y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(classification_report(y_test , y_pred))
```
