```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
    'I love to play football and basketball.',
    'I enjoy watching movies and reading books.',
    'This is the fourth document.',
    'Imran Khan is the former Prime Minister of Pakistan.',
    'Saeedullah is a software engineer and data scientist.',
    'I love to have tea and coffee in the morning.',
]
```

```python
v = TfidfVectorizer()
```

```python
x = v.fit_transform(corpus)
```

```python
print(v.vocabulary_)
```

```python
all_features = v.get_feature_names_out()
```

```python
all_features #This is object its not a list so we can not use index on it. We can convert it to list and then use index on it.
```

```python
for word in all_features:
    indx = v.vocabulary_.get(word) #IT contain every single word and its index in the vocabulary
    print(f"{word} : {v.idf_[indx]}") #Print the IDF value for each word
```
