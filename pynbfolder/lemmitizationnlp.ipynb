```python
import nltk
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words = ["going" , "eaten" , "working" , "running" , "bad boy" , "good" , "funny" , "What" , "joke" , "history"]
```

```python
for word in words:
    print(f"{word} -------> {lemmatizer.lemmatize(word , pos='v')}")
```
