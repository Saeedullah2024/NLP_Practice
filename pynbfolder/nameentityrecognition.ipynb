```python
sentence = "Saeed traveled from Karachi to Dubai in August 2026 to attend a conference organized by Microsoft."
```

```python
import nltk

nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from  nltk.chunk import ne_chunk
```

```python
words = word_tokenize(sentence)
```

```python
words
```

```python
wordstag = pos_tag(words)
```

```python
wordstag
```

```python
name_entity_recog = ne_chunk(wordstag)
```

```python
name_entity_recog
```

```python
entities = []

for entity in name_entity_recog:
    if hasattr(entity, "label"):
            entity_text = " ".join(word for word, tag in entity)
    entities.append({
                "text": entity_text,
                "label": entity.label()
            })
```
