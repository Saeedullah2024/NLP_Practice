```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stopwords
```

```python
stopwordlist = stopwords.words('english')
```

```python
stopwordlist
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
```

```python
sentences = sent_tokenize(paragraph)
wordtoken = word_tokenize(paragraph)
```

```python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in stopwordlist]
    sentences[i] = ' '.join(words)
    
```

```python
# Stopwords ---> Filter ---> Stemming
```

```python
from nltk.stem import SnowballStemmer
```

```python
Snowstemmer = SnowballStemmer(language='english')
```

```python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [Snowstemmer.stem(word) for word in words if word not in stopwordlist]
    sentences[i] = ' '.join(words)
    
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
wordlemm = WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [wordlemm.lemmatize(word , pos = 'v') for word in words if word not in stopwordlist]
    sentences[i] = ' '.join(words)
    
```
