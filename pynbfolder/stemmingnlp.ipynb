```python
words = ["Saeed" , "writes" , "writing", "an" , "essay" , "on" , "a" , "written" , "topic" , "called" , "The Pakistan" , "Topic"]
```

```python
words
```

```python
#The PorterStemmer
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
for word in words:
    print(f"{word} ------> {stemmer.stem(word)}")
```

```python
from nltk.stem import RegexpStemmer
```

```python
regexstem = RegexpStemmer('ing$|s$|es$|able$' , min=4)
```

```python
for word in words:
    print(f"{word} ------> {regexstem.stem(word)}")
```
