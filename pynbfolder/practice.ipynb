```python
from nltk.stem import PorterStemmer
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
words = ["run", "running", "ran", "easily", "fairly"]
```

```python
import nltk
```

```python
nltk.download('wordnet')
```

```python
for word in words:
    print(f"Porter stemmer for {word} : {PorterStemmer().stem(word)}")
    print(f"Word net stemmer for {word} : {WordNetLemmatizer().lemmatize(word , pos='v')}")
```
